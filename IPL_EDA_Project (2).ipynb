```python
top_batsmen = (
    df.groupby("batsman")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_batsmen
```

```python
plt.figure(figsize=(10,6))

top_batsmen.plot(kind="bar")

plt.title("Top 10 Batsmen by Total Runs")
plt.xlabel("Batsman")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)
plt.show()
```

### Insight

The chart identifies the highest run-scoring batsmen in the dataset.
It allows us to compare the overall batting contribution of the leading
players.

```python
bowler_wicket_types = [
    "caught",
    "bowled",
    "lbw",
    "stumped",
    "caught and bowled",
    "hit wicket"
]

wickets = df[df["dismissal_kind"].isin(bowler_wicket_types)]

top_bowlers = (
    wickets.groupby("bowler")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

top_bowlers
```

```python
plt.figure(figsize=(10,6))

top_bowlers.plot(kind="bar")

plt.title("Top 10 Bowlers by Wickets")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.xticks(rotation=45)
plt.show()
```

```python
team_runs = (
    df.groupby("batting_team")["total_runs"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

team_runs.plot(kind="bar")

plt.title("Total Runs Scored by Teams")
plt.xlabel("Team")
plt.ylabel("Runs")
plt.xticks(rotation=70)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(x="total_runs", data=df)

plt.title("Distribution of Runs per Delivery")
plt.xlabel("Runs")
plt.ylabel("Number of Deliveries")
plt.show()
```

```python
boundaries = df[df["batsman_runs"].isin([4,6])]["batsman_runs"].value_counts()

boundaries.plot(kind="bar")

plt.title("Number of Fours and Sixes")
plt.xlabel("Boundary Type")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

```python
six_hitters = (
    df[df["batsman_runs"] == 6]
    .groupby("batsman")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

six_hitters.plot(kind="bar")

plt.title("Top 10 Batsmen with Most Sixes")
plt.xlabel("Batsman")
plt.ylabel("Number of Sixes")
plt.xticks(rotation=45)
plt.show()
```

```python
dismissals = df[
    df["dismissal_kind"] != "No Dismissal"
]["dismissal_kind"].value_counts()

plt.figure(figsize=(10,6))

dismissals.plot(kind="bar")

plt.title("Types of Dismissals")
plt.xlabel("Dismissal Type")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.show()
```

```python
runs_by_over = df.groupby("over")["total_runs"].sum()

plt.figure(figsize=(10,5))

plt.plot(runs_by_over.index, runs_by_over.values, marker="o")

plt.title("Total Runs Scored by Over")
plt.xlabel("Over")
plt.ylabel("Total Runs")
plt.xticks(range(1,21))
plt.grid(True)
plt.show()
```

```python
extras = df[
    [
        "wide_runs",
        "bye_runs",
        "legbye_runs",
        "noball_runs",
        "penalty_runs"
    ]
].sum()

plt.figure(figsize=(9,5))

extras.plot(kind="bar")

plt.title("Distribution of Extra Runs")
plt.xlabel("Extra Type")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)
plt.show()
```

```python
numeric_columns = [
    "wide_runs",
    "bye_runs",
    "legbye_runs",
    "noball_runs",
    "penalty_runs",
    "batsman_runs",
    "extra_runs",
    "total_runs"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["batsman_runs"])

plt.title("Boxplot of Batsman Runs per Delivery")
plt.xlabel("Runs")
plt.show()
```

```python
top_fielders = (
    df[df["dismissal_kind"] == "caught"]["fielder"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,6))

top_fielders.plot(kind="bar")

plt.title("Top 10 Fielders by Catches")
plt.xlabel("Fielder")
plt.ylabel("Catches")
plt.xticks(rotation=45)
plt.show()
```
